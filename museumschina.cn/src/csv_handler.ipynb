```python
import pandas as pd
from pathlib import Path
BASE_DIR = Path.cwd().parent.parent
# print(BASE_DIR)
INPUT_CSV = BASE_DIR / r"museumschina.cn\data\pottery_details_20260305_210613.csv"
df = pd.read_csv(INPUT_CSV)
```

```python
# 展示列名，大小
print(df.columns)
print(df.shape)
```

```python
# 去重
df.drop_duplicates(inplace=True)
print(df.shape)
```

```python
# 通过 data wrangler 发现有一行缺失gallery_urls，找到这一行
missing_gallery_urls = df[df['gallery_urls'].isnull()]
print(missing_gallery_urls)
# 实际访问网页发现这件文物真的没有轮播图，且主图也无法访问，说明数据无法使用，删除这一行
df.dropna(subset=['gallery_urls'], inplace=True)
print(df.shape)
```

```python
# 输出为含时间戳的新csv
from datetime import datetime
TIME_STAMP = datetime.now().strftime("%Y%m%d_%H%M%S")
OUTPUT_CSV = BASE_DIR / rf"museumschina.cn\data\data_cleaned_{TIME_STAMP}.csv"
df.to_csv(OUTPUT_CSV, index=False)
```
