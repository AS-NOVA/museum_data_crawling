## 2. 根据name填写period

现在已经将所需的数据转为了一个dataframe，在内存中名为`new_df`。需要在本笔记本，即`museumschina.cn\src\data_to_pps_csv.ipynb`中编写新的单元格，根据`name`这一列，提取时代`era`和时期`period`信息，并填入相应名称的列中。例如，对于“新石器时代仰韶文化夹砂红陶缸”，这里称其时代`era`为`新石器时代`，时期`period`为`仰韶文化`。  

对`era`和`period`的确定应遵守如下规则：  
- 如果无法从名称中看出`era`，应默认填写`新石器时代`。在没有进一步的`period`的条件下，本字段允许进行粗略的时间分期，例如如果名称中仅提及`新石器时代早期`且未提及文化，则应将`新石器时代早期`作为`era`。
- 如果无法从名称中看出`period`，应当留空不填。本字段允许包含文化的分期信息，即如果名称中提及`xx文化早期/中期/晚期`，则允许将其作为`period`。
- 如果出现了什么以上规则没有明确涉及的问题，可以考虑允许大模型将其写入一个额外的备注字段。

为清晰起见，此步骤应分两阶段编写代码和执行：先调用大模型api（从deepseek官网购买），读取`new_df`的`name`列，并输出完整的结构化的json数据，其中至少应包含`era`和`period`两个字段，保存到本地`museumschina.cn\data`文件夹中。然后，再从该文件夹中读取数据，填入`new_df`中。这样能够将大模型的输出与后续处理分离，避免重复执行此笔记本代码时重新调用api，造成金钱浪费。

```python
import os
import json
import logging
import pandas as pd
from pathlib import Path
from datetime import datetime
from tqdm.auto import tqdm
from openai import OpenAI
from dotenv import load_dotenv

# 1. 路径配置 (严格遵循 Pathlib 和 BASE_DIR 规范)
DATA_DIR = BASE_DIR / "museumschina.cn" / "data"
LOG_DIR = BASE_DIR / "museumschina.cn" / "log"
ENV_PATH = BASE_DIR / "museumschina.cn" / ".env"  # 指定 .env 路径

# 动态生成带时间戳的文件路径
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
OUTPUT_JSONL_PATH = DATA_DIR / f"extracted_era_period_{timestamp}.jsonl"
LOG_FILE_PATH = LOG_DIR / f"api_extraction_{timestamp}.log"

# 2. 日志记录配置
logging.basicConfig(
    filename=LOG_FILE_PATH,
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s',
    encoding='utf-8'
)

# 2. 加载环境变量
if ENV_PATH.exists():
    load_dotenv(dotenv_path=ENV_PATH)
    logging.info("成功加载 .env 配置文件")
else:
    # 强制报错，防止在没有 Key 的情况下运行浪费时间
    raise FileNotFoundError(f"未在 {BASE_DIR} 找到 .env 文件，请先创建并配置 API Key")

# 3. 初始化客户端 (从环境变量读取)
client = OpenAI(
    api_key=os.getenv("DEEPSEEK_API_KEY"),
    base_url=os.getenv("DEEPSEEK_BASE_URL", "https://api.deepseek.com")
)

print(f"✅ 路径初始化完成。日志将写入: {LOG_FILE_PATH.name}")
print(f"✅ 提取结果将持续写入: {OUTPUT_JSONL_PATH.name}")
```

```python
# 准备 Prompt 规则
SYSTEM_PROMPT = """你是一位考古学专家，负责从彩陶名称中提取时间信息。
请根据提供的彩陶名称，按要求返回结构化的 JSON 数据。

返回字段的规则：
1. era (时代): 
   - 默认值为“新石器时代”。
   - 若名称中包含“新石器时代早期/中期/晚期”，则提取完整描述。
2. period (时期/文化): 
   - 提取文化名称（如“仰韶文化”、“马家窑文化”）。
   - 允许包含分期信息（如“仰韶文化晚期”）。
   - 若名称中未提及具体文化，请留空串 ""。

输出格式要求：
必须只返回 JSON 对象，格式如下：
{"era": "...", "period": "..."}
"""

def extract_info(name):
    try:
        response = client.chat.completions.create(
            model="deepseek-chat",
            messages=[
                {"role": "system", "content": SYSTEM_PROMPT},
                {"role": "user", "content": f"请从以下名称中提取信息：{name}"},
            ],
            response_format={'type': 'json_object'},
            stream=False
        )
        content = response.choices[0].message.content
        logging.info(f"SUCCESS - Name: {name}")
        return json.loads(content)
    except Exception as e:
        logging.error(f"FAILED - Name: {name} - Error: {str(e)}")
        return {"era": "新石器时代", "period": "", "remark": f"Error: {str(e)}"}

# 执行提取（逐条写入 JSONL，避免等待全部完成后再落盘）
results = {}
with open(OUTPUT_JSONL_PATH, 'w', encoding='utf-8') as f:
    # 使用 tqdm 追踪进度，new_df 为内存中已有的 DataFrame
    for index, row in tqdm(new_df.iterrows(), total=len(new_df), desc="LLM 提取进度"):
        pottery_name = row['name']
        extracted = extract_info(pottery_name)

        # 保留内存结果，便于后续直接回填 DataFrame
        results[index] = extracted

        # 实时追加一行 JSON，包含索引和名称，便于后续对齐
        record = {
            "index": int(index),
            "name": pottery_name,
            "era": extracted.get("era", "新石器时代"),
            "period": extracted.get("period", ""),
            "remark": extracted.get("remark", "")
        }
        f.write(json.dumps(record, ensure_ascii=False) + "\n")
        f.flush()

print(f"🎉 第一阶段完成！提取结果已保存至: {OUTPUT_JSONL_PATH}")
```

## 4. 根据images_url填写images_extension，还可考虑填写images_description
简单的字符串处理  
由于彩陶书数据路径类似于images/1/xxx.jpg，而当前数据路径类似images/xxx.jpg。美观期间，均改为images/museumchina/xxx.jpg再提交上去，但本地路径先不改。

```python
import os

def extract_extensions(path_str):
    if pd.isna(path_str) or str(path_str).strip() == "":
        return None
    paths = str(path_str).split('|')
    # Extract extension (e.g., jpg) - remove dot
    exts = [os.path.splitext(p)[1].lstrip('.') for p in paths]
    return "|".join(exts)

def modify_image_path(path_str):
    if pd.isna(path_str) or str(path_str).strip() == "":
        return None
    paths = str(path_str).split('|')
    # Replace 'images/' with 'images/museumchina/'
    new_paths = [p.replace('images/', 'images/museumchina/', 1) if p.startswith('images/') else p for p in paths]
    return "|".join(new_paths)

# Ensure data consistency by resetting from source
new_df['images_url'] = df['all_images_paths']

# 1. 提取扩展名到 images_extension (不带点)
new_df['images_extension'] = new_df['images_url'].apply(extract_extensions)
   
# 2. 修改 images_url 中的路径前缀
new_df['images_url'] = new_df['images_url'].apply(modify_image_path)

# 显示结果验证
new_df.head()
```
