```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path.cwd().parent
DATA_DIR = BASE_DIR / "data"
EXTRACTED_JSONL = DATA_DIR / "extracted" / "extracted_metadata_20260402_104557.jsonl"
ORIGIN_CSV = DATA_DIR / "data_cleaned.csv"
```

```python
df_extracted = pd.read_json(EXTRACTED_JSONL, lines=True)
df_origin = pd.read_csv(ORIGIN_CSV)
merged_df = pd.merge(df_origin, df_extracted, left_on="id", right_on="source_id", how="inner")
merged_df = merged_df.drop(columns=["source_id"])
```

```python
merged_df.columns
```

```python
merged_df.drop(columns=["pictureIds","imgUrl","yearName","original_name"],inplace=True)
merged_df = merged_df[[
    'id', 'name',
    'era', 'culture', 'decoration', 'shape_feature','texture', 'root_shape',
    'condition',
    'size', 'description', 'collectionTexture',
    'museumName', 'local_image_paths']]
merged_df.head(5)
```

```python
shape_feature_with_root = merged_df[['shape_feature', 'root_shape']].value_counts().reset_index(name='频数')
shape_feature_with_root.head(20)
```

```python
shape_feature_to_root = pd.crosstab(merged_df['shape_feature'], merged_df['root_shape'], normalize='index')
shape_feature_to_root_pct = (shape_feature_to_root * 100).round(2)
shape_feature_to_root_pct.loc["高柄"].sort_values(ascending=False)[lambda x: x > 0]
```
